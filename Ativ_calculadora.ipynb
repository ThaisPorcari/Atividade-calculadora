```python
#Calculadora - precisa ter os botões:

#Somar
def somar(a, b):
  somar = a + b
  return print(f"A soma dos números {a} + {b} é igual a {somar}")

#Subtrair
def subtrair (a, b):
    subtrair = a - b
    return print(f"A diferença dos números {a} - {b} é igual a {subtrair}")

#Multiplicar
def multiplicar (a, b):
  multiplicar = a * b
  return print(f"A multiplicação dos números {a} x {b} é igual a {multiplicar}")

#Dividir
def dividir (a, b):
  if b == 0:
    return print (f"Erro: Não é possivel dividir um número por zero")
  else:
    dividir = a / b
    return print(f"A divisão dos números {a} / {b} é igual a {dividir}")

while True:
  print("\n --- CALCULADORA PYTHON ---")
  print ("1 - Somar ")
  print ("2 - Subtrair ")
  print ("3 - Multiplicar ")
  print ("4 - Dividir ")
  print ("0 - Sair ")

  escolha = input("Escolha uma opção: ")

  if escolha == "0":
    print(" Calculadora encerrada")
    break
  if escolha in ["1","2","3","4"]:
    num1 = float(input(f"Informe o primeiro número: "))
    num2 = float(input(f"Informe  segundo  número: "))

    if escolha == "1":
      resultado = somar (num1 , num2)

    elif escolha == "2":
      subtrair (num1 , num2)

    elif escolha == "3":
      resultado = multiplicar (num1 , num2)

    elif escolha == "4":
      resultado = dividir (num1 , num2)

  else:
    print("Escolha um número válido")





5

```
